```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Assuming the file is in the current working directory of your Colab notebook
data = pd.read_csv('student_learning_habits_dataset.csv')

# Display the first few rows of the dataframe to check if it loaded correctly
data.head()


# Get a summary of the data
data.info()

# Check for any missing values
data.isnull().sum()
```

```python
# 1. Age Distribution (Histogram)
plt.figure(figsize=(10, 6))
sns.histplot(data['Age'], bins=10, kde=True, color="skyblue")
plt.title("Age Distribution of Students")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# 2. Preferred Learning Environment (Bar Chart)
plt.figure(figsize=(10, 6))
sns.countplot(data['Preferred Learning Environment'], palette="viridis")
plt.title("Preferred Learning Environment")
plt.xlabel("Learning Environment")
plt.ylabel("Count")
plt.show()
```

```python

# 3. Study Hours/Week (Pie Chart)
study_hours_counts = data['Study Hours/Week'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(study_hours_counts, labels=study_hours_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette("pastel"))
plt.title("Study Hours/Week Distribution")
plt.show()
```

```python

# 4. Study Satisfaction by Learning Style (Box Plot)
plt.figure(figsize=(10, 6))
sns.boxplot(x='Learning Style', y='Study Satisfaction', data=data, palette="coolwarm")
plt.title("Study Satisfaction by Learning Style")
plt.xlabel("Learning Style")
plt.ylabel("Study Satisfaction")
plt.show()
```

```python

# 5. Primary Device Usage (Bar Chart)
plt.figure(figsize=(10, 6))
sns.countplot(data['Primary Device'], palette="Set2")
plt.title("Primary Device Usage")
plt.xlabel("Device")
plt.ylabel("Count")
plt.show()
```

```python
# Ensure the Seaborn style
sns.set(style="whitegrid")

# 1. Gender Distribution in Preferred Learning Environments (Clustered Bar Chart)
plt.figure(figsize=(10, 6))
sns.countplot(x='Preferred Learning Environment', hue='Gender', data=data, palette="coolwarm")
plt.title("Gender Distribution in Preferred Learning Environments")
plt.xlabel("Preferred Learning Environment")
plt.ylabel("Count")
plt.legend(title="Gender")
plt.show()
```

```python
# 2. Study Hours vs. Study Satisfaction (Box Plot)
plt.figure(figsize=(10, 6))
sns.boxplot(x='Study Hours/Week', y='Study Satisfaction', data=data, palette="Blues")
plt.title("Study Hours vs. Study Satisfaction")
plt.xlabel("Study Hours per Week")
plt.ylabel("Study Satisfaction")
plt.show()
```

```python
# 3. Learning Style by Study Time (Heatmap)
# Create a crosstab to count occurrences of each combination of Learning Style and Study Time
learning_time_crosstab = pd.crosstab(data['Learning Style'], data['Study Time'])
plt.figure(figsize=(10, 6))
sns.heatmap(learning_time_crosstab, annot=True, cmap="YlGnBu", fmt="d")
plt.title("Learning Style by Study Time")
plt.xlabel("Study Time")
plt.ylabel("Learning Style")
plt.show()
```

```python

# 4. Primary Device and Study Tools Preference (Stacked Bar Chart)
# Create a crosstab for Primary Device and Study Tools
device_tool_crosstab = pd.crosstab(data['Primary Device'], data['Study Tools'])
device_tool_crosstab.plot(kind="bar", stacked=True, figsize=(10, 6), colormap="Set3")
plt.title("Primary Device and Study Tools Preference")
plt.xlabel("Primary Device")
plt.ylabel("Count")
plt.legend(title="Study Tools")
plt.show()
```

```python
# 5. Correlation Matrix (Heatmap for numerical variables)
# Select only numerical columns for correlation
numerical_data = data.select_dtypes(include=['int64', 'float64'])
plt.figure(figsize=(8, 6))
sns.heatmap(numerical_data.corr(), annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Matrix")
plt.show()
```
