```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# Read and prepare the data
df = pd.read_csv('student_learning_habits_Analysis(Responses) - Form responses 1.csv')
# Remove 'Column 1' and 'Column 12' as they're not needed
df = df.drop(['Column 1', 'Column 12'], axis=1)

```

```python
# Print basic summary information
print("=== Dataset Summary ===")
print(f"\nTotal number of responses: {len(df)}")
print("\nMissing Values:")
print(df.isnull().sum())

print("\n=== Numerical Summary ===")
print(df[['Age', 'Study Satisfaction']].describe())

print("\n=== Categorical Distributions ===")
for col in ['Gender:', 'Preferred Learning Environment', 'Study Time', 'Learning Style']:
    print(f"\n{col} Distribution:")
    print(df[col].value_counts())
```

```python
# 1. Gender Distribution (Pie Chart)
plt.subplot(3, 2, 1)
gender_counts = df['Gender:'].value_counts()
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Gender Distribution of Students', pad=20)
```

```python
#  2. Learning Environment by Satisfaction (Bar Plot)
plt.subplot(3, 2, 2)
env_satisfaction = df.groupby('Preferred Learning Environment')['Study Satisfaction'].mean()
sns.barplot(x=env_satisfaction.index, y=env_satisfaction.values)
plt.title('Average Study Satisfaction by Learning Environment')
plt.xticks(rotation=45)
plt.xlabel('Average Satisfaction Score')
```

```python
# 3. Learning Style Distribution (Bar Plot)
plt.subplot(3, 2, 3)
learning_style_counts = df['Learning Style'].value_counts()
sns.barplot(x=learning_style_counts.index, y=learning_style_counts.values)
plt.xticks(rotation=45)
plt.title('Distribution of Learning Styles')
plt.ylabel('Count')
```

```python
# 4. Study Hours Distribution by Gender (Stacked Bar Chart)
study_hours_gender = pd.crosstab(df['Study Hours/Week'], df['Gender:'])
study_hours_gender.plot(kind='bar', stacked=True)
plt.title('Study Hours Distribution by Gender')
plt.xlabel('Study Hours per Week')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.legend(title='Gender')
```

```python
# 5. Primary Device Usage (Horizontal Bar Plot)
plt.subplot(3, 2, 5)
device_counts = df['Primary Device'].value_counts()
sns.barplot(y=device_counts.index, x=device_counts.values, orient='h')
plt.title('Primary Device Usage for Studying')
plt.xlabel('Count')
```

```python
# 6. Study Time Preferences (Count Plot)
plt.subplot(3, 2, 6)
sns.countplot(data=df, y='Study Time', order=df['Study Time'].value_counts().index)
plt.title('Preferred Study Time Distribution')
plt.xlabel('Count')

plt.tight_layout(pad=3.0)
# plt.suptitle('Student Learning Habits Analysis - Part 1', fontsize=16, y=1.02)
plt.show()
```

```python
#  Additional visualizations
# 7. Age Distribution
plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
sns.histplot(data=df, x='Age', bins=15)
plt.title('Age Distribution of Students')
plt.xlabel('Age')
plt.ylabel('Count')
```

```python
#  Create a bar plot
plt.figure(figsize=(10, 6))
df['Study Tools'].value_counts().plot(kind='bar')

# Customize the plot
plt.title('Distribution of Study Tools')
plt.xlabel('Study Tools')
plt.ylabel('Number of Students')
plt.xticks(rotation=45, ha='right')
plt.grid(True, axis='y')

# Adjust layout
plt.tight_layout()

# Show plot
plt.show()
```
